```julia
using Plots
using LinearAlgebra
using DataInterpolations
```

# Linear Splines

## Using DataInterpolations.jl

```julia
t = [0., 1.5, 2, 3.1, 4, 5.2, 6.];
y = sin.(t);
scatter(t, y, label="Data")
```

For `LinearInterpolation` note the order of the arguments:

```julia
itp = LinearInterpolation(y,t)
```

```julia
xx = LinRange(0,6,500);
plot(xx, sin.(xx), label="sin(x)")
plot!(xx, itp.(xx), label="Linear Interpolation")
scatter!(t, y, label="Data")
xlabel!("x")
```

### Extrapolation
By default, `LinearInterpolation` does **not** extrapolate.  We need to specify extrapolation (if we want it).

```julia
itp_ex = LinearInterpolation(y,t, 
    extrapolation_left=ExtrapolationType.Extension, 
    extrapolation_right=ExtrapolationType.Extension)
```

`Extension` says we just keep using the leftmost/rightmost element

```julia
xx = LinRange(0,2*pi,500);
plot(xx, sin.(xx), label="sin(x)")
plot!(xx, itp_ex.(xx), label="Linear Interpolation")
scatter!(t, y, label="Data")
xlabel!("x")
```

## Interpolations Version

```julia
using Interpolations
```

Note the order of arguments is reversed:

```julia
itp2 = linear_interpolation(t, y)
# itp2 = linear_interpolation(t, y, extrapolation_bc=Line())
```

```julia
xx = LinRange(0,6,500);
plot(xx, sin.(xx), label="sin(x)")
plot!(xx, itp2.(xx), label="Linear Interpolation")
scatter!(t, y, label="Data")
xlabel!("x")
```

# Cubic Splines

## DataInterpolations Version
Defaults to natural boundary conditions, $$S''(t_0) = S''(t_n)=0$$.

```julia
t = [0., 1.5, 2, 3.1, 4, 5.2, 6.];
y = sin.(t);
linear_itp = LinearInterpolation(y,t);
cubic_itp = CubicSpline(y,t);

xx = LinRange(0,6,500);
plot(xx, sin.(xx), label="sin(x)")
plot!(xx, linear_itp.(xx), label="Linear Interpolation")
plot!(xx, cubic_itp.(xx), label="Cubic Spline")
scatter!(t, y, label="Data")
xlabel!("x")
```

## Interpolations Version
* Defaults to Natural BCs
* Requires a regular (i.e., uniform) set of knots

```julia
using Interpolations
```

```julia
t = t = 0:1:6;
y = sin.(t);
linear_itp2 = linear_interpolation(t,y);
cubic_itp2 = cubic_spline_interpolation(t,y);

xx = LinRange(0,6,500);
plot(xx, sin.(xx), label="sin(x)")
plot!(xx, linear_itp2.(xx), label="Linear Interpolation")
plot!(xx, cubic_itp2.(xx), label="Cubic Spline")
scatter!(t, y, label="Data")
xlabel!("x")
```

# Splines vs. Lagrange

```julia
n = 1000;
t = LinRange(0,2*pi,n+1);
y = sin.(t);
lagrange_itp = LagrangeInterpolation(y,t);
cubic_itp = CubicSpline(y,t);

xx = LinRange(0,2*pi,500);
plot(xx, sin.(xx), label="sin(x)")
plot!(xx, lagrange_itp.(xx), label="LagrangeInterpolant")
plot!(xx, cubic_itp.(xx), label="Cubic Spline")
scatter!(t, y, label="Data")
xlabel!("x")
title!("n = $n")
```

Lagrange went ``bad'' at $n=100$.

```julia
plot(xx, sin.(xx), label="sin(x)")
# plot!(xx, lagrange_itp.(xx), label="LagrangeInterpolant")
plot!(xx, cubic_itp.(xx), label="Cubic Spline")
scatter!(t, y, label="Data")
xlabel!("x")
title!("n = $n")
```

Cubic splines are much more robust than Lagrange interpolation
