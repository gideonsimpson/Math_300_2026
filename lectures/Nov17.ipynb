# Example 1

```julia
f(x) = sin(x);
h_vals = [0.1, 0.01, 0.001, 0.0001]
x=1.;
err_vals = Float64[]
for h in h_vals
    # compute approximate derivative using finite difference
    df = (f(x + h) - f(x)) / h
    # compute error
    err = abs(df - cos(x))
    push!(err_vals, err);
    println("h: $h, Approximate derivative: $df, error = $(err)")
end
```

```julia
scatter(h_vals, err_vals, 
xscale=:log10, yscale=:log10, 
 xlabel="h", ylabel="",  
title="Error in Numerical Differentiation", label="Error" )
plot!(h_vals, h_vals, label="∝ h", color=:red, linestyle=:dash)
```

We see the $\mathrm{O}(h)$ error in the approximation

# Example 1 Revisited
Make $h$ very very small

```julia
f(x) = sin(x);
h_vals = 10.0 .^(-1:-1:-12)
x=1.;
err_vals = Float64[]
for h in h_vals
    # compute approximate derivative using finite difference
    df = (f(x + h) - f(x)) / h
    # compute error
    err = abs(df - cos(x))
    push!(err_vals, err);
    # println("h: $h, Approximate derivative: $df, error = $(err)")
end

scatter(h_vals, err_vals, 
xscale=:log10, yscale=:log10, 
 xlabel="h", ylabel="",  
title="Error in Numerical Differentiation", label="Error" )
plot!(h_vals, h_vals, label="∝ h", color=:red, linestyle=:dash)
xticks!(10.0 .^(-1:-1:-12))
yticks!(10.0 .^(-1:-1:-12))
```

U-shaped error.  There is an optimal $h_\star$ such that if $h< h_\star$, we get a worse estimate.  We no longer follow the $\propto h$ trend line.  This a generic problem for this approximation.

# Higher Order Methods

```julia
f(x) = sin(x);
h_vals = 10.0 .^(-1:-1:-12)
x=1.;
err_vals = Float64[]
for h in h_vals
    # compute approximate derivative using finite difference
    df = (f(x + h) - f(x-h)) / (2h)
    # compute error
    err = abs(df - cos(x))
    push!(err_vals, err);
    # println("h: $h, Approximate derivative: $df, error = $(err)")
end

scatter(h_vals, err_vals, 
xscale=:log10, yscale=:log10, 
 xlabel="h", ylabel="",  
title="Error in Centered Difference Scheme", label="Error" , legend=:topleft)
plot!(h_vals, h_vals, label="∝ h", color=:red, linestyle=:dash)
xticks!(10.0 .^(-1:-1:-12))
yticks!(10.0 .^(-1:-1:-12))
```

Least squares fitting of the slope

This tells us $p\approx 2$.  The 95\% confidence interval includes 2.
