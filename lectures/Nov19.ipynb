```julia
using Plots
```

The $\varphi$ function:

```julia
function centered_difference(f, x, h)
    return (f(x + h) - f(x - h)) / (2 * h)
end
```

For estimating $f'(1)$ where $f(x) = \sin(x)$:

```julia
f(x) = sin(x);
h_vals = 10.0 .^(-1:-1:-12)
x=1.;
err_vals = Float64[]
for h in h_vals
    # compute approximate derivative using finite difference
    df = centered_difference(f, x, h)
    # compute error
    err = abs(df - cos(x))
    push!(err_vals, err);
    # println("h: $h, Approximate derivative: $df, error = $(err)")
end

scatter(h_vals, err_vals, 
xscale=:log10, yscale=:log10, 
 xlabel="h", ylabel="",  
title="Error in Centered Difference Scheme", label="Error" , legend=:topleft)
plot!(h_vals, h_vals.^2, label="∝ h²", color=:red, linestyle=:dash)
xticks!(10.0 .^(-1:-1:-12))
yticks!(10.0 .^(-1:-1:-12))
ylims!(1e-12, 1e-1)
```

The $\psi$  function:

```julia
centered_diff_richardson(f, x, h) = (4 * centered_difference(f, x, h/2) - centered_difference(f, x, h)) / 3
```

```julia
f(x) = sin(x);
h_vals = 10.0 .^(-1:-1:-12)
x=1.;
err_vals = Float64[]
for h in h_vals
    # compute approximate derivative using finite difference
    df = centered_diff_richardson(f, x, h)
    # compute error
    err = abs(df - cos(x))
    push!(err_vals, err);
    # println("h: $h, Approximate derivative: $df, error = $(err)")
end

scatter(h_vals, err_vals, 
xscale=:log10, yscale=:log10, 
 xlabel="h", ylabel="",  
title="Error in Centered Difference Scheme", label="Error" , legend=:topleft)
plot!(h_vals, h_vals.^4, label="∝ h⁴", color=:red, linestyle=:dash)
xticks!(10.0 .^(-1:-1:-14))
yticks!(10.0 .^(-1:-1:-14))
ylims!(1e-14, 1e-1)
```
