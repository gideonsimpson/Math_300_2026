# Composite Trapezoidal Rule

```julia
function comp_trapz(x_nodes, f_nodes)
    integral = 0.0;
    n = length(x_nodes)-1;
    for i in 1:n
        h = x_nodes[i+1] - x_nodes[i];
        integral += h * (f_nodes[i] + f_nodes[i+1]) / 2;
    end
    return integral;
end
```

```julia
a = 0;
b = 2;
h = b-a;

f(x) = x^2 -3x + 2; #quadratic function
@show integral_exact = 2/3;
@show integral_trapz = h * (f(a) + f(b)) / 2;
n = 100;
x_nodes = range(a, b, length=n+1);
f_nodes = f.(x_nodes);
@show integral_comp_trapz = comp_trapz(x_nodes, f_nodes);
```

```julia
f(x) = sin(x);
a = 0;
b = 2;
integral_exact = 2*sin(1)^2; #integral over [0,2]

n_vals = [4, 8, 16, 32, 64, 128];
estimates = Float64[];
for n in n_vals
    x_nodes = LinRange(a,b,n+1);
    f_nodes = f.(x_nodes);
    integral_comp_trapz =  comp_trapz(x_nodes, f_nodes);
    push!(estimates, integral_comp_trapz);
end

scatter((b-a)./n_vals, abs.(estimates .- integral_exact), label="Data", xlabel="h", ylabel="Error",xscale=:log2,yscale=:log2)
plot!((b-a)./n_vals, .01 * (b-a)./n_vals, label="∝ h¹", color=:red,ls=:dash)
plot!((b-a)./n_vals, .1 * ((b-a)./n_vals).^2, label="∝ h²", color=:orange,ls=:dash)
xticks!(2. .^(-8:-1))
yticks!(2. .^(-15:-4))
```

Appears to be a quadratic method, $\mathrm{O}(h^2)= \mathrm{O}(1/n^2)$
