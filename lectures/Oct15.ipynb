# Fixed Point Example
$$
F(x) = 4 + \tfrac{1}{3}\sin(2x)
$$

```julia
x0 = 1.0;
F(x) = 4 + 1/3*  sin(2*x);

println("Iteration $(0): x = $x0")

for i in 1:10
    x0 = F(x0)
    println("Iteration $i: x = $x0")
end
```

```julia
xx = LinRange(0, 2π, 100)
plot(xx, F.(xx), label="F(x)", xlabel="x", ylabel="F(x)")
plot!(xx, xx, label="y=x", linestyle=:dash)
```

If $y=F(x)$ intercepts $y=x$, that corresponds to a fixed point of $F$.

# Problem Statement
Find all roots of
$$
z^3 -1 = 0
$$
via Newton's method.

 As a cubic, there are three roots, roots of unity, and these can be obtained by writing 
$$
z = re^{i\theta}
$$
and $1 = e^{i 2n\pi}$ for any $n \in \mathbb{Z}$, to get
$$
r^3 e^{i 3 \theta} =e^{i 2n\pi},
$$
to get that the roots are
$$
e^{i \theta_n}, \quad \theta_n = \frac{2n\pi}{3}.
$$
The three unique ones are
$$
w_1 = 1,\quad w_2 = e^{i2\pi/3}, \quad w_3 =  e^{i4\pi/3}.
$$
These are visuralized in the complex plane in the following plot:

```julia
roots = [1 + 0im, cis(2π/3), cis(4π/3)]
scatter(roots,aspect_ratio=:equal,label="Roots")
xlabel!("Real Part")
ylabel!("Imaginary Part")
θ = LinRange(0, 2π, 100)
circle_x = cos.(θ)
circle_y = sin.(θ)
plot!(circle_x, circle_y, label="Unit Circle", linewidth=2, color=:red,ls=:dash)
xlims!(-1.5,1.5)
ylims!(-1.5,1.5)
```

For this problem, $f(z) = z^3 -1$ and Newton's method corresponds to
$$
z_{n+1} = z_n - \frac{z_n^3 -1}{3z_n^2}
$$
Given a starting point, $z_0$, we will see which of the three roots, $w_1$, $w_2$, and $w_3$, converges $z_n$ converges to. 

```julia
f(z) = z^3 - 1
fp(z) = 3z^2

function find_root(z0; tol=1e-6, maxiter=1000)
    z = z0
    for n in 1:maxiter
        z_new = z - f(z)/fp(z)
        if abs(z_new - z) < tol
            z = z_new;
            break
        end
        z = z_new
        if(n == maxiter)
            @warn z0
            @warn "Max iterations reached without convergence"
        end
    end
    idx = argmin([abs(z - r) for r in roots]);
    return idx
end



```

The following plot shows, by color, which of the $w_1$, $w_2$, $w_3$, Newton converges to at the $z_0$ in the complex plane at which we start Newton.

```julia
x = LinRange(-2, 2, 400);
y = LinRange(-2, 2, 400);
z = [xi + im*yi for yi in y, xi in x];
root_ids = find_root.(z);
heatmap(x, y, root_ids, legend=false, clims=(1, 3), colorbar=true,size=(800, 800),
    color=:viridis, xlabel="Re z", ylabel="Im z", xlims=(-2, 2), ylims=(-2, 2))
    scatter!(roots,label="Roots")
```

This a "Newton Fractal"
