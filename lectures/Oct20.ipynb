# Lagrange

```julia
x_nodes = [1, 1.5, 3, 3.5, 5]; # n = 4 here
```

```julia
@show k = 2; # which l_{k} we construct
@show n = length(x_nodes)-1; # degree of interpolant

xx = LinRange(1, 5, 100); # plot between 1 and 5
Lnk = ones(length(xx));

# textbook indexing in j and j+1 for Julia
for j in 0:k-1
    @. Lnk *= (xx - x_nodes[j+1])/(x_nodes[k+1] -  x_nodes[j+1])
end

for j in k+1:n
    @. Lnk *= (xx - x_nodes[j+1])/(x_nodes[k+1] -  x_nodes[j+1])
end

plot(xx, Lnk, label="l_{k}", legend=:topright)
scatter!(x_nodes, zeros(n + 1), label="Data", color=:orange)
scatter!([x_nodes[k+1]], [1], label="", color=:red)
hline!([0,1], color=:black, ls=:dash,label="")
xlabel!("x")
title!("k = $k")
```
