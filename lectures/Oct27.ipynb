```julia
using Plots
using DataInterpolations
using LinearAlgebra
```

# Manually Construct Lagrange Interpolant for $n=2$

```julia
x_nodes = [1., 3., 5.];

f = x-> 1/x;

L0 = x-> (x^2 - 8*x + 15)/8;
L1 = x-> -(x^2 - 6*x + 5)/4;
L2 = x-> (x^2 - 4*x + 3)/8;

# note that Julia indexes arrays starting at 1
P= x-> f(x_nodes[1])*L0(x) + f(x_nodes[2])*L1(x) + f(x_nodes[3])*L2(x);

xx = LinRange(1,5,100);

plot(xx, f.(xx), label="f(x)")
plot!(xx, P.(xx),label="Interpolant")
scatter!(x_nodes, f.(x_nodes),label="Data")
xlabel!("x")
```

# Interpolating using `DataInterpolations.jl`

```julia
x_nodes
```

To use `LagrangeInterpolation` we pass the arrays as `LagrangeInterpolation(y,x)`

```julia
p_pkg = LagrangeInterpolation(f.(x_nodes), x_nodes)
```

```julia
xx = LinRange(1,5,100);

plot(xx, f.(xx), label="f(x)")
plot!(xx, P.(xx),label="Manually Built Interpolant")
plot!(xx, p_pkg.(xx),label="Package Interpolant")
scatter!(x_nodes, f.(x_nodes),label="Data")
xlabel!("x")
```

```julia
x_eval = LinRange(1,5,500);
@show norm(P.(x_eval) - p_pkg.(x_eval)); # computes the ℓ₂ norm of the difference
```

# Interpolation Going Wrong

```julia
x_nodes = LinRange(0,1,100);
f= x-> sin(x); y_nodes = f.(x_nodes)
xx = LinRange(0,1,5000);

p = LagrangeInterpolation(y_nodes, x_nodes)
println(norm(f.(x_nodes) - p.(x_nodes),Inf));
plot(xx, f.(xx), label="f(x)")
plot!(xx, p.(xx), label="Interpolant")
scatter!(x_nodes, y_nodes, label="Data")
xlabel!("x")
ylabel!("y")
title!("Error of Interpolation = $(norm(f.(xx) - p.(xx),Inf))")
```

In general, Lagrange interpolation becomes unstable for uniformly spaced nodes, as $n$ gets large; related to Vandermonde matrix ill-conditioning.  This is also a sign of **overfitting** from statistics.  This is called the *Runge* phenomenon.
