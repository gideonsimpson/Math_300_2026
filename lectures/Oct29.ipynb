````julia
using Plots
using DataInterpolations
using LinearAlgebra
````

# Chebyshev Nodes
Chebyshev nodes do not dengerate as $n\to \infty$; uniformly spaced nodes do.

````julia

n = 65;
x_unif = LinRange(-1,1,n); 
x_cheb = sort([cos(pi * (2 * i+1)/(2*n+2)) for i in 0:n]);
# # y_nodes = @. sin(2* π * x_nodes);
y_unif = @. exp(2* π * x_unif);
y_cheb = @. exp(2* π * x_cheb);
# 
p_unif = LagrangeInterpolation(y_unif, x_unif, extrapolation_left=ExtrapolationType.Extension, extrapolation_right=ExtrapolationType.Extension); 
p_cheb = LagrangeInterpolation(y_cheb, x_cheb, extrapolation_left=ExtrapolationType.Extension, extrapolation_right=ExtrapolationType.Extension); 

xx = LinRange(-1,1,5000);
plot(xx, p_unif.(xx),label="Uniform Interpolant")
plot!(xx, p_cheb.(xx),label="Chebyshev Interpolant")
scatter!(x_unif, y_unif, label="Uniform Data", alpha =0.5)
scatter!(x_cheb, y_cheb, label="Chebyshev Data", alpha =0.5)
title!("n = $(n)")
xlabel!("x")
````

Understanding 
```
extrapolation_left=ExtrapolationType.Extension, extrapolation_right=ExtrapolationType.Extension
```
When we evaluate poitns outside of the minimum and maximum values of the nodes, we are **extrapolating**.  We need to tell the code how it should extrapolate.

````julia
n =7;
@show x_unif = LinRange(-1,1,n)|>collect; 
@show x_cheb = sort([cos(pi * (2 * i+1)/(2*n+2)) for i in 0:n]);
````

````julia
n = 5;
x_cheb = sort([cos(pi * (2 * i+1)/(2*n+2)) for i in 0:n]);
y_cheb = @. exp(2* π * x_cheb);
# 
# p_cheb = LagrangeInterpolation(y_cheb, x_cheb); # no extrapolation sepcified. 
p_cheb = LagrangeInterpolation(y_cheb, x_cheb, extrapolation_left=ExtrapolationType.Extension, extrapolation_right=ExtrapolationType.Extension); 

# xx = LinRange(minimum(x_cheb),maximum(x_cheb),5000); # only plots where we have data

xx = LinRange(-1,1,5000); # tries to plot beyond where we have data.  


plot(xx, p_cheb.(xx),label="Chebyshev Interpolant")
scatter!(x_cheb, y_cheb, label="Chebyshev Data", alpha =0.5)
````

To evaluate byeond where we have data, we need to tell it how to extrapolate.  `ExtrapolationType.Extension` tells it to just evaluate the interpolant beyond where we have data. Most interpolations packages ask you to specify an extrapolation rule.
