# Taylor Examples

## Example 1
$$
f(x) = e^x
$$

```julia
xx = LinRange(-2,2,100);
f(x) = exp(x);

c = 0.5;

P0(x) = f(c);
P1(x) = f(c) + f(c) * (x - c);
P2(x) = f(c) + f(c) * (x - c) + f(c) / 2 * (x - c)^2;
P3(x) = f(c) + f(c) * (x - c) + f(c) / 2 * (x - c)^2 + f(c) / 6 * (x - c)^3;


plot(xx, f.(xx), label="f(x)")
plot!(xx, P0.(xx), label="P0(x)")
plot!(xx, P1.(xx), label="P1(x)")
plot!(xx, P2.(xx), label="P2(x)")
plot!(xx, P3.(xx), label="P3(x)")
title!("c = $c")
```

## Example 2
$$
f(x) = \ln(1+x)
$$

```julia
xx = LinRange(0, 1, 100);
f(x) = log(1+x);
fp(x) = 1 / (1 + x);
fpp(x) = -1 / (1 + x)^2;
c = 0;

P0(x) = f(c);
P1(x) = f(c) + fp(c) * (x - c);
P2(x) = f(c) + fp(c) * (x - c) + fpp(c) / 2 * (x - c)^2;

plot(xx, f.(xx), label="f(x)")
plot!(xx, P0.(xx), label="P0(x)")
plot!(xx, P1.(xx), label="P1(x)")
plot!(xx, P2.(xx), label="P2(x)")
title!("c = $c")
```

## Example 3
$$
f(x) = \sqrt{1+x}
$$

```julia
xx = LinRange(0, 1, 100);
f(x) = sqrt(1+x);
fp(x) = 0.5 / sqrt(1 + x);
fpp(x) = -0.25 / (1 + x)^1.5;
c = 0;

P0(x) = f(c);
P1(x) = f(c) + fp(c) * (x - c);
P2(x) = f(c) + fp(c) * (x - c) + fpp(c) / 2 * (x - c)^2;

plot(xx, f.(xx), label="f(x)")
plot!(xx, P0.(xx), label="P0(x)")
plot!(xx, P1.(xx), label="P1(x)")
plot!(xx, P2.(xx), label="P2(x)")
title!("c = $c")
```

# Newton's Method

```julia
function newton_method(f, df, x0; nmax = 100, δ=1e-6, ϵ = 1e-6)
    x = x0;
    for n in 1:nmax
        fx = f(x);
        dfx = df(x);
        if abs(fx) < ϵ
            @info("Newton converged to root after $n iterations -- small magnitude of f(x)")
            return x
        end
        x_new = x - fx / dfx;
        if abs(x_new - x) < δ
            @info("Newton converged to root after $n iterations -- small change in x")
            return x_new
        end
        x = x_new;
    end
    @error("Newton did not converge after $nmax iterations")
    return x
end
```

## Example
$$
f(x) = \exp(x) - \sin(x)
$$

```julia
f(x) = exp(x) - sin(x);
df(x) = exp(x) - cos(x);
xx = LinRange(-7, 1, 100);
plot(xx, f.(xx))
hline!([0])
```

```julia
newton_method(f, df, -3)
```

```julia
newton_method(f, df, -6)
```

## Avoiding Derivatives
Use automatic differentiation

```julia
using ForwardDiff # provides automatic differentiation tools
using LinearAlgebra # provides linear algebra tools, in cluding the norm function
```

```julia
df_auto(x) = ForwardDiff.derivative(f, x);
df_auto(0.5)
```

```julia
xx = LinRange(-7, 1, 100);
plot(xx, df.(xx),label="Analytic Derivative")
plot!(xx, df_auto.(xx), label="Automatic Derivative")
xlabel!("x")
```

```julia
@show norm(df.(xx) - df_auto.(xx));
```

Excellent agreement in values

```julia
newton_method(f, df, -6)
```

```julia
newton_method(f, df_auto, -6)
```

```julia
using BenchmarkTools
```

```julia
@btime df.($xx);
@btime df_auto.($xx);
```

Automatic differentiation, after initialized, is very close to analytic differentiation, in terms of performance.
