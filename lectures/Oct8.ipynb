```julia
function newton_method(f, df, x0; nmax = 100, δ=1e-6, ϵ = 1e-6)
    x = x0;
    for n in 1:nmax
        fx = f(x);
        dfx = df(x);
        if abs(fx) < ϵ
            @info("Newton converged to root after $n iterations -- small magnitude of f(x)")
            return x
        end
        x_new = x - fx / dfx;
        if abs(x_new - x) < δ
            @info("Newton converged to root after $n iterations -- small change in x")
            return x_new
        end
        x = x_new;
    end
    @error("Newton did not converge after $nmax iterations")
    return x
end
```

# Implicit Example
$$
G(x,y)= y^5 + y + x^2 -1=0
$$

```julia
G(x,y) = y^5 + y +x^2 -1;
dGdy(x,y) = 5*y^4 + 1;

x_vals = LinRange(-4,4,100);
y_vals = Float64[]
for x in x_vals
    y0 = 0.0;
    # x is frozen in the following Newton iteration
    y = newton_method(y -> G(x,y), y -> dGdy(x,y), y0)
    # println("x = $x, y = $y")
    push!(y_vals, y)    
end
```

```julia
plot(x_vals, y_vals, label="y(x)")
```
