```julia
function secant_method(f, x0, x1; nmax = 100, δ=1e-6, ϵ = 1e-6)
    x = x0;
    for n in 1:nmax
        f0 = f(x0);
        f1 = f(x1);
        if abs(f1) < ϵ
            @info("Secant converged to root after $n iterations -- small magnitude of f(x)")
            return x1
        end

        if abs(x1-x0) < δ
            @info("Secant converged to root after $n iterations -- small change in x")
            return x1
        end

        x = x - f1 * (x1-x0)/(f1 - f0);
        x0 = x1;
        x1 = x;
    end
    @error("Secant did not converge after $nmax iterations")
    return x1
end
```

```julia
f(x) = cos(x+1)- x;

xx = LinRange(-1, 1, 100);
plot(xx, f.(xx), label="f(x)", xlabel="x", ylabel="f(x)", title="Secant Method Example")
```

```julia
secant_method(f, -1, 1)
```

## Multidiimensional Root Finding

```julia
using NonlinearSolve
```

Given $F(x,y) = (F_1(x,y), F_2(x,y))$, 
\begin{align*}
F_1(x,y)& = x^2 +y^2 -4\\
F_2(x,y)& = e^x -y
\end{align*}
solve $F(x,y) = (0,0)$

```julia
x = range(-2.5, 1.5, length=300)
y = range(-0.5, 2.5, length=300)

F1(x, y) = x^2 + y^2 - 4
F2(x, y) = exp(x) - y

contour(x, y, F1,
    levels=-5:1:10,
    linewidth=3,
    label="F₁(x,y) = 0",
    aspect_ratio=:equal,cmap=:RdBu)

contour!(x, y, F2,
    levels=-5:1:10,
    linewidth=3,
    label="F₂(x,y) = 0",cmap=:RdBu)

xlabel!("x")
ylabel!("y")
```

```julia
# define F = (F1, F2)
function F(u, p)
    x, y = u # u is a 2-vector, and I am extracting the two components into x and y

    return [
        x^2 + y^2 - 4   # computes F1
        exp(x) - y      # computes F2
    ]
end
```

The `p` is short for parameters to allow us to handle more general problems, with variable constants, i.e.:
\begin{align*}
F_1(x,y)& = a\cdot x^2 +b \cdot y^2 -4\\
F_2(x,y)& = e^x -y
\end{align*}
and `p = (a,b)`.  Julia expects your `F` to be coded to accept an optional `p` value, even if it's trivial.



```julia
u0 = [-1.0, 1.0]

prob = NonlinearProblem(F, u0)
sol = solve(prob)

sol.u
```

```julia
x = range(-2.5, 1.5, length=300)
y = range(-0.5, 2.5, length=300)

F1(x, y) = x^2 + y^2 - 4
F2(x, y) = exp(x) - y

contour(x, y, F1,
    levels=-5:1:10,
    linewidth=3,
    label="F₁(x,y) = 0",
    aspect_ratio=:equal,cmap=:RdBu)

contour!(x, y, F2,
    levels=-5:1:10,
    linewidth=3,
    label="F₂(x,y) = 0",cmap=:RdBu)
scatter!([sol.u[1]], [sol.u[2]], color=:red, label="solution")
xlabel!("x")
ylabel!("y")
```
