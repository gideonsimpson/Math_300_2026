# Defining Functions

```julia
# compact one-line function definition
f(x,y) = x^2 + y^2;

# anonymous function stored in a variable
g = (x,y) -> x^2 + y^2;

# standard multi-line function definition
function h(x,y)
    return x^2 + y^2  # same calculation: x^2 + y^2
end
```

```julia
@show f(1,2)
@show g(1,2)
@show h(1,2)
```

## Arrays and Vectors
In Julia, 1D arrays correspend to vectors.  Julia vectors can be created using `[` and `]` , and **indexing begins at 1**.

```julia
u = [1, 2, 3]
```

```julia
u[1]
```

```julia
u[2:3]
```

```julia
u[end]
```

```julia
v = [1., 2, 3]
```

By default, these are column vectors.  Get row vectors with a `'` tranpose operation:

```julia
v'
```
