# Example of a sequence with particular limit:
$$
x_n = \frac{1}{n}
$$
So $\lim_{n\to \infty} x_n = 0$.

```julia
nmax = 10;
for n in 1:nmax
    xn = 1/n;
    println("n = $n, xn = $xn");
end
```

We would need $n > 100$ to even get two significant figures right.

# Example 
$$
x_n = \left(1 + \frac{1}{n}\right)^n
$$

$x_n \to e=2.71828...$

```julia
nmax = 100;
for n in 1:nmax
    xn = (1+1/n)^n;
    println("n = $n, xn = $xn");
end
```

# Fast Convergence Example
$$
x_{n+1} = \frac{1}{2} x_n + \frac{1}{x_n}, \quad x_1 = 2
$$

```julia
nmax = 10;
xn = 2;
for n in 1:nmax
    xn = .5 * xn + 1/xn;
    println("n = $n, xn = $xn");
end
```

```julia
@show sqrt(2);
```

This converged quickly

# Comparing Sequences

```julia
nmax = 8;
xn = [1/n for n in 1:nmax];
yn = [(1/3^n) for n in 1:nmax];
zn = [1/(2^(2^n)) for n in 1:nmax];
```

In place modification of an existing plot uses commands with `!`:

Also note the use of the log scaling

```julia
scatter(1:nmax, xn, label="xn",yaxis=(:log10, [1e-15, :auto])) # plots xn
scatter!(1:nmax, yn, label="yn")
scatter!(1:nmax, zn, label="zn")
xlabel!("n")
```

* $y_n$ is linearly convergent to zero
* $z_n$ is superlinearly convergent to zero
* $x_n$ is sublinear
